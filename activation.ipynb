```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sig(x):
  y=1/(1+np.exp(-x))
  return y
```

```python
x = np.linspace(-10, 10, 100)
y = sig(x)

plt.figure(figsize=(4, 3))
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('sig(x)')
plt.title('Sigmoid Function')
plt.grid(True)
plt.show()
```

```python
def tanh(x):
  y=(2/(1+np.exp(-2*x)))-1
  return y
y1 = tanh(x)

plt.figure(figsize=(4, 3))
plt.plot(x, y1)
plt.xlabel('x')
plt.ylabel('tanh(x)')
plt.title('tanh Function')
plt.grid(True)
plt.show()
```

```python
def relu(x):
  return np.maximum(0, x)


y2 = relu(x)

plt.figure(figsize=(4, 3))
plt.plot(x, y2)
plt.xlabel('x')
plt.ylabel('relu(x)')
plt.title('relu Function')
plt.grid(True)
plt.show()
```

```python
def softmax(x, axis=-1):
    e_x = np.exp(x - np.max(x, axis=axis, keepdims=True))
    return e_x / e_x.sum(axis=axis, keepdims=True)
```

```python
x_softmax = np.linspace(-5, 5, 100)
# Softmax typically operates on a vector, so for plotting a single input
# we can consider a 2-element vector where one element is x_softmax and the other is 0
# or simply demonstrate its properties with a single input conceptually.
# For a single scalar input, softmax is just 1. So, let's create a small array for demonstration.

almost_zero = 1e-10 # To avoid division by zero if all elements are negative
data_for_softmax_plot = np.array([x_softmax, np.zeros_like(x_softmax) + almost_zero])

# Apply softmax along the rows (axis=0) to get two arrays, each corresponding to an input
y_softmax = softmax(data_for_softmax_plot)

plt.figure(figsize=(4, 3))
plt.plot(x_softmax, y_softmax[0], label='softmax(x, 0)')
plt.xlabel('x')
plt.ylabel('softmax output')
plt.title('Softmax Function Example')
plt.grid(True)
plt.legend()
plt.show()
```
